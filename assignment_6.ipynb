```python

import gymnasium as gym

def random_agent(env_name, episodes=5):
    env = gym.make(env_name)
    print("\nEnvironment:", env_name)

    total_rewards = []

    for episode in range(1, episodes + 1):
        state, info = env.reset()
        total_reward = 0
        steps = 0

        while steps < 2000:
            # Select a random action
            action = env.action_space.sample()

            # Perform the action
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1
            state = next_state

            # Stop when episode ends
            if terminated or truncated:
                break

        average_reward = total_reward / steps
        total_rewards.append(total_reward)

        print(f"\nEpisode {episode}")
        print("Total Reward:", total_reward)
        print("Steps:", steps)
        print("Average Reward:", round(average_reward, 2))

    print("\nOverall Average Episode Reward:",
          round(sum(total_rewards) / episodes, 2))

    env.close()


# Run both environments
random_agent("CliffWalking-v1", episodes=5)
random_agent("Taxi-v4", episodes=5)
```
