```python
# Policy Iteration in Deterministic 2x2 Grid World

# 1. Define parameters
gamma = 0.9
theta = 0.001
normal_reward = -1
wall_reward = -1
goal_reward = 100

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
terminal_state = 'G'

# 2. Define transitions: (next_state, reward)
transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}

# 3. Initialize policy and state values
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

iteration = 0

print("INITIAL POLICY:", policy)

# 4. Policy Iteration
while True:

    iteration += 1
    print("\n================================")
    print("POLICY ITERATION:", iteration)
    print("================================")

    # POLICY EVALUATION
    while True:
        delta = 0
        new_V = V.copy()

        for state in states:
            action = policy[state]
            next_state, reward = transitions[state][action]

            value = reward + gamma * V[next_state]

            delta = max(delta, abs(value - V[state]))
            new_V[state] = value

        V = new_V

        print("Policy Evaluation:", V)

        if delta < theta:
            break

    # ACTION VALUE CALCULATION
    Q = {}

    for state in states:
        Q[state] = {}

        for action in actions:
            next_state, reward = transitions[state][action]

            Q[state][action] = reward + gamma * V[next_state]

    print("\nAction Values Q:")
    for state in states:
        print(state, Q[state])

    # POLICY IMPROVEMENT
    new_policy = {}

    for state in states:
        best_action = max(Q[state], key=Q[state].get)
        new_policy[state] = best_action

    print("\nImproved Policy:", new_policy)

    # POLICY STABILITY CHECK
    if new_policy == policy:
        print("\nPolicy is stable!")
        break

    policy = new_policy


# 5. Final Results
print("\n========== FINAL RESULTS ==========")

for state in states:
    print(state, "Value =", round(V[state], 3),
          "Optimal Action =", policy[state])

print("G Value =", V['G'], "(Terminal)")


# 6. Final Q Value Table
print("\n========== FINAL Q VALUE TABLE ==========")

for state in states:
    print("\n", state)
    for action in actions:
        print(action, ":", round(Q[state][action], 3))


# 7. Bellman Optimality Verification
print("\n========== BELLMAN OPTIMALITY ==========")

for state in states:
    max_q = max(Q[state].values())

    print(state,
          "V*(s) =", round(V[state], 3),
          "Max Q*(s,a) =", round(max_q, 3))


# 8. Optimal Path
print("\nOptimal Path: S0 -> S1 -> G")
```
