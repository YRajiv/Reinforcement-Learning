```python

import random

# Part A: Initialize machines
machines = ['A', 'B', 'C']
epsilon = 0.2
n_plays = 10

# Part B: Initialize N, R and Q
N = {m: 0 for m in machines}
R = {m: [] for m in machines}
Q = {m: 0.0 for m in machines}

total_reward = 0

# Part C: Epsilon-Greedy Strategy
for play in range(1, n_plays + 1):

    r = random.random()

    if r < epsilon:
        # Exploration: select a random machine
        machine = random.choice(machines)
        strategy = "Exploration"
    else:
        # Exploitation: select the machine with highest Q
        best_q = max(Q.values())
        best_machines = [m for m in machines if Q[m] == best_q]
        machine = random.choice(best_machines)
        strategy = "Exploitation"

    print(f"\nPlay {play}")
    print("Strategy:", strategy)
    print("Selected machine:", machine)

    # Part D: Collect reward
    reward = float(input("Enter reward obtained: "))

    # Part E: Update N, R and Q
    N[machine] += 1
    R[machine].append(reward)

    Q[machine] = Q[machine] + (
        reward - Q[machine]
    ) / N[machine]

    total_reward += reward

    print("Updated Q-values:", {
        m: round(Q[m], 2) for m in machines
    })

# Part F: Display final results
print("\n========== FINAL RESULTS ==========")

for m in machines:
    print(f"\nMachine {m}")
    print("Participation count N:", N[m])
    print("Rewards:", R[m])
    print("Estimated value Q:", round(Q[m], 2))

print("\nTotal reward after 10 plays:", total_reward)
```
