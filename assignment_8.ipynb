```python
%%writefile grid_env_det.py

ROWS = 2
COLS = 3

GAMMA = 0.9

REWARDS = {
    "normal": -1,
    "goal": 10,
    "obstacle": -5
}

ACTIONS = {
    "Up": (-1, 0),
    "Down": (1, 0),
    "Left": (0, -1),
    "Right": (0, 1)
}

START = "S0"

STATES = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G": (0, 2),
    "X": (1, 0),
    "S3": (1, 1),
    "S4": (1, 2)
}

print("Grid World environment loaded successfully!")
```

```python
%%writefile assignment1.py

import grid_env_det as env

GAMMA = env.GAMMA
STATES = env.STATES
ACTIONS = env.ACTIONS

POLICY = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

def transition(state, action):
    row, col = STATES[state]
    dr, dc = ACTIONS[action]

    nr, nc = row + dr, col + dc

    # Stay in the same state if outside the grid
    if not (0 <= nr < env.ROWS and 0 <= nc < env.COLS):
        return state, env.REWARDS["normal"]

    next_state = next(
        s for s, pos in STATES.items() if pos == (nr, nc)
    )

    if next_state == "X":
        return next_state, env.REWARDS["obstacle"]

    if next_state == "G":
        return next_state, env.REWARDS["goal"]

    return next_state, env.REWARDS["normal"]


def calculate_return(state, first_action=None):
    total = 0
    discount = 1.0

    if first_action is not None:
        state, reward = transition(state, first_action)
        total += discount * reward

        if state in ("G", "X"):
            return total

        discount *= GAMMA

    while state not in ("G", "X"):
        action = POLICY[state]
        state, reward = transition(state, action)

        total += discount * reward
        discount *= GAMMA

    return total


def state_value_function(state):
    return calculate_return(state)


def action_value_function(state, action):
    return calculate_return(state, first_action=action)


print("STATE VALUES")
for state in ["S0", "S1", "S3", "S4"]:
    print(f"V({state}) = {state_value_function(state):.3f}")

print("\nACTION VALUE")
print(f"Q(S0, Up) = {action_value_function('S0', 'Up'):.3f}")
```
