```python

# Bellman Optimality and Value Iteration

gamma = 0.9
theta = 0.001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

# Transition: next state and reward
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 10)
    }
}

V = {"S0": 0, "S1": 0, "G": 0}

print("Initial Values:", V)

# Value Iteration
iteration = 1

while True:
    new_V = V.copy()

    for state in states:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            values.append(q)

        new_V[state] = max(values)

    delta = max(abs(new_V[s] - V[s]) for s in states)

    V = new_V

    print("Iteration", iteration, "Values:", V,
          "Delta:", round(delta, 3))

    iteration += 1

    if delta < theta:
        break

# Calculate Q-values and optimal policy
print("\nFinal State Values:", V)

for state in states:
    q_values = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        q_values[action] = reward + gamma * V[next_state]

    max_q = max(q_values.values())

    best_actions = [
        action for action in actions
        if abs(q_values[action] - max_q) < 0.000001
    ]

    print("\nState:", state)
    print("Q-values:", q_values)
    print("Optimal Action(s):", best_actions)

    # Verify Bellman Optimality
    print("V*(s):", V[state])
    print("Maximum Q-value:", max_q)
```
